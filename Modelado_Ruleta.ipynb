# La Ruleta, el modelo

Como dijimos, primero lo primero. Entender el problema.

Vamos a modelar un juego de ruleta. Para ello debemos conocer bien las reglas del juego:
* Hay 37 números: 36 números (18 $\color{red}{\textbf{rojos}}$ y 18 $\color{black}{\textbf{negros}}$), más el 0 ($\color{green}{\textbf{negro}}$).

* Hay varios tipos de apuestas: Vamos a considerar sólo la sencilla (__un número__). Donde:

    1) Si ganamos, nos dan 35 veces lo que apostamos.

    2) Si perdemos, perdemos lo que apostamos.

## Elementos del juego.

```python
# El casino
1- Una ruleta:
    La ruleta representa un número al azar que va desde 0 hasta 36 inclusive.

# Elecciones del apostador
2- Elección de un número:
    Imaginemos que somos una persona muy supersticiosa. Solo vamos a elegir un número
   "n_suerte", que es nuestro número de la suerte y por nada en el mundo lo cambiaremos.

3- Dinero que perder... o ganar:
    Para poder jugar debemos modelar el importe que apostaremos por "apuesta" y  un importe
    total o sea un "capital". Imaginemos que contamos con un capital de $5.000 y vamos a
    jugar de $50 por vez ("apuesta").
```

````python
import random

capital = 5000
apuesta = 50
n_suerte = int(input("Elegí un número entre 0 y 36: "))

resultados = []
tiradas = 0

while capital >= apuesta and tiradas < 100:
    ruleta = random.randint(0, 36)
    tiradas = tiradas + 1
    if ruleta == n_suerte:
        capital = capital + 35 * apuesta
        print("¡Ganaste! Salió el", ruleta)
    else:
        capital = capital - apuesta
        print("Perdiste. Salió el", ruleta)
    resultados.append(capital)

print("Cantidad de tiradas:", tiradas)
print("Capital final:", capital)
print("Ganancia o pérdida:", capital - 5000)
print("Historial del capital:", resultados)
````

````python
import random

capital = 5000
apuesta = 50
n_suerte = int(input("Elegí un número entre 0 y 36: "))

# Colores de la ruleta
rojos = [1, 3, 5, 7, 9, 12, 14, 16, 18, 19, 21, 23, 25, 27, 30, 32, 34, 36]
negros = [2, 4, 6, 8, 10, 11, 13, 15, 17, 20, 22, 24, 26, 28, 29, 31, 33, 35]

resultados = []
tiradas = 0

while capital >= apuesta and tiradas < 100:
    ruleta = random.randint(0, 36)
    tiradas = tiradas + 1
    if ruleta in rojos:
        color = "rojo"
    elif ruleta in negros:
        color = "negro"
    else:
        color = "verde"

    if ruleta == n_suerte:
        capital = capital + 35 * apuesta
        print("¡Ganaste! Salió el", ruleta, color)
    else:
        capital = capital - apuesta
        print("Perdiste. Salió el", ruleta, color)

    resultados.append(capital)

print("Cantidad de tiradas:", tiradas)
print("Capital final:", capital)
print("Ganancia o pérdida:", capital - 5000)
print("Historial del capital:", resultados)
````

## La dinámica de la apuesta

El juego consiste en unos pocos y simples pasos. Elegir un número de la ruleta (0-36) antes de que el croupier de la ruleta cante "_No va más_" y se disponga a tirar la bolilla.

Los pasos para jugar son:

    0- Contar con un "capital"
    1- Elegir un número ("n_suerte")
    2- Realizar una "apuesta"
    3- Tirar "ruleta"
    4- actualizar el "capital"

##Juguemos varias veces
Ahora probemos jugar unas 100 veces. Siempre apostamos al mismo número y vamos actualizando el capital.
Para eso vamos a necesitar hacer una lista donde ir almacenando el capital luego de cada jugada.
Si el jugador queda con capital cero o menos, ya no puede seguir jugando (Ayuda: hoy vimos como poner dos condicionales al mismo tiempo)

````python
import random

dinero = int(input("Con cuanto dinero ingresas?"))
n_suerte = 8
dinero_apostado = int(input("cuanto apostas?"))

for n in range(100):
    if dinero >= dinero_apostado:
      numero = random.randint(0,36)
      if numero == n_suerte:
        dinero_ganado = dinero_apostado *35
        dinero = dinero + dinero_ganado
        print(f"Ganaste: {dinero_ganado}")
      else:
        dinero = dinero - dinero_apostado
        print(f"Perdiste tu dinero y te quedan:{dinero} pesos")


if dinero <= 0: # este if está fuera del for
  print('No tenés más plata para apostar, andate a tu casa')
````

````python
import random

capital_inicial = int(input("🪙​Cuanto Dinero tienes?"))
capital = capital_inicial
apuesta = int(input("💲​​Cuanto apuestas?"))
n_suerte = int(input("Elegí un número entre 0 y 36: "))
tiradas = 0
numeros_rojos = {1, 3, 5, 7, 9, 12, 14, 16, 18, 19, 21, 23, 25, 27, 30, 32, 34, 36}

while capital >= apuesta and tiradas < 100:
    ruleta = random.randint(0, 36)
    tiradas = tiradas + 1
    if ruleta == 0:
        color = "verde 🟢"
    elif ruleta in numeros_rojos:
        color = "rojo 🔴"
    else:
        color = "negro ⚫"

    if ruleta == n_suerte:
        capital = capital + 35 * apuesta
        print(f"¡Ganaste! Salió el {ruleta} {color}")
    else:
        capital = capital - apuesta
        print(f"Perdiste. Salió el {ruleta} {color}")
    if capital < apuesta:
        print("​💔​Te quedaste sin dinero ¡Fin del juego!​💔​")
        break

diferencia = capital - capital_inicial
verbo = "​‼️​ganado" if diferencia >= 0 else "❌​perdido"
print(f"Con un total de {tiradas} tiradas te quedaste con un capital de 💲{capital}, habiendo {verbo} ${abs(diferencia)}")
````
